```python
item1=int(input("Enter the cost of first item: "))
item2=int(input("Enter the cost of secand item: "))
item3=int(input("Enter the cost of third item: "))
a=int(input("Enter the quantity of first item: "))
b=int(input("Enter the quantity of secand item: "))
c=int(input("Enter the quantity of secand item: "))
total_cost=a*item1 + b*item2 + c*item3
if (total_cost>50) :
  discount=total_cost-((total_cost*10)/100)
  print("Total cost after discount is: " ,discount)
else :
    print("The total cost is: " ,total_cost)
```
